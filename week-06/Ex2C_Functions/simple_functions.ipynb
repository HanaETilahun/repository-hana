This is Lab 1 of Exercise 2.C. I am working with defining custom functions.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician is {music}."

print(favorite_things("Hana", "Black Panther", "Teddy Afro"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined Year Up United Data Analyst program to build my data skills and prepare for a strong internship."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Ethiopia", "it feels like home and connects me to my family and culture"))
```

```python
print(favorite_place("Dubai", "it is beautiful, modern, and full of exciting places to visit"))
```
