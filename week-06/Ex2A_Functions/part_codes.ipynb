# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier(part_code):
    colon_index = part_code.find(":")
    supplier = part_code[:colon_index]
    return supplier


def get_product_code(part_code):
    colon_index = part_code.find(":")
    dash_index = part_code.find("-")
    product_code = part_code[colon_index + 1:dash_index]
    return product_code


def get_size(part_code):
    dash_index = part_code.find("-")
    size = part_code[dash_index + 1:]
    return size


def parse_part_code(part_code):
    supplier = get_supplier(part_code)
    product_code = get_product_code(part_code)
    size = get_size(part_code)

    print(f"Supplier: {supplier}")
    print(f"Product Code: {product_code}")
    print(f"Size: {size}")


parse_part_code("ACME:123-L")
parse_part_code("DI:12345-M")
parse_part_code("ACE:1-12")
```

```python
# Function to get supplier code (before the :)
def get_supplier_code(part_code):
    return part_code.split(":")[0]

# Test
print(get_supplier_code("ACME:123-L"))
```

```python
# Function to get product number (between : and -)
def get_product_num(part_code):
    return part_code.split(":")[1].split("-")[0]

# Test
print(get_product_num("ACME:123-L"))
```

```python
# Function to get size (after the -)
def get_size(part_code):
    return part_code.split("-")[1]

# Test
print(get_size("ACME:123-L"))
```

```python
# Variables holding the example part codes
part1 = "ACME:123-L"
part2 = "DI:12345-M"
part3 = "ACE:1-12"
```
