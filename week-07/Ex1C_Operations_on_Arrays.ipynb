```python
import numpy as np
```

## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.rand(4, 5)

print(r_1)
print(r_2)
print(r_3)
print(r_4)
print(r_5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The shuffle changed the order of the rows in r_4. It changed the original array directly.

## Mathematical Operations

```python
print(r_1)
print(r_1.shape)

print(r_2)
print(r_2.shape)

result = np.add(r_1, r_2)
print(result)
print(result.shape)
```

```python
print(r_1)
print(r_1.shape)

print(r_3)
print(r_3.shape)

result = np.add(r_1, r_3)
print(result)
print(result.shape)
```

```python
print(r_2)
print(r_2.shape)

print(r_4)
print(r_4.shape)

result = np.add(r_2, r_4)
print(result)
print(result.shape)
```

```python
# r_1 + r_4 gives an error because the shapes do not match for broadcasting.
# r_2 + r_4 works because r_2 has 1 row and 2 columns only if the columns match.
# Since r_4 has 5 columns, r_2 + r_4 gives an error.
# r_3 + r_4 gives an error because r_3 has 2 rows and 1 column, but r_4 has 4 rows and 5 columns.

print(np.add(r_1, r_4))
```

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))
```

```python
# This gives an error because [[1], [1]] has 2 rows,
# but r_4 has 4 rows. The shapes cannot broadcast together.

print(np.add(np.array([[1], [1]]), r_4))
```

Broadcasting lets NumPy combine arrays with different shapes when the smaller array can fit across the bigger array. The add function adds matching values together. If one array has only one value, that value can be added to every value in the larger array. The shape of the final result usually follows the larger array.

## Grocery Problem

```python
# Order: vegetables, fruits, snacks, beverages

bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
category_totals = np.add(bag1, bag2)

print(f"There are {category_totals[0]} vegetables, {category_totals[1]} fruits, {category_totals[2]} boxes of snacks, and {category_totals[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
ben = np.array([0, 5, 5, 3, 4])
carla = np.array([3, 6, 0, 3, 4])
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
person_totals = np.sum(all_knocks, axis=1)
group_total = np.sum(all_knocks)

print(f"Ali knocked on {person_totals[0]} doors.")
print(f"Ben knocked on {person_totals[1]} doors.")
print(f"Carla knocked on {person_totals[2]} doors.")
print(f"The group knocked on {group_total} doors total.")
daily_totals = np.sum(all_knocks, axis=0)
rolling_total = np.cumsum(daily_totals)
estimated_votes = rolling_total * 0.5

print(estimated_votes)
```
