# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
        
```

The first loop multiplies each row by 2 and prints the row as an array.

The second loop goes through each individual number and multiplies each number by 2 separately.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This repeats the whole row multiplied by 2
# because x represents the full row, not the single value y.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# This gives an error because the arrays have different shapes.
# NumPy requires the same shape when stacking arrays.

stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# This gives an error because the arrays do not have the same number of rows.
# axis=1 joins side-by-side, so both arrays must match in row count.

another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)
```

```python
h_stack = np.hstack((arrA, arrA))  # Combines arrays side by side
v_stack = np.vstack((arrA, arrA))  # Combines arrays top to bottom
d_stack = np.dstack((arrA, arrA))  # Combines arrays into depth layers

print(h_stack)
print(v_stack)
print(d_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# The result is 2 arrays.
# Each array is 2-dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)

# The result is 3 arrays.
# Each array is 2-dimensional.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
