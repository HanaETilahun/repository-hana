# Exercise 3.B More plotting practice

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Advanced Plotting Techniques

```python
products = ["Product A", "Product B", "Product C", "Product D"]
sales = [23, 17, 35, 29]

plt.bar(products, sales)

plt.show()
```

```python
plt.bar(products, sales, color="skyblue", width=0.5)

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)

plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
s = np.random.rand(50) * 100

plt.scatter(x, y, s=s, c=s, cmap="viridis")

plt.title("Random Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

## Subplots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

random_numbers = np.random.randn(1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(x, y)
axes[0].set_title("Sine Wave")
axes[0].set_xlabel("X Values")
axes[0].set_ylabel("sin(x)")

axes[1].hist(random_numbers)
axes[1].set_title("Histogram of Random Numbers")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
